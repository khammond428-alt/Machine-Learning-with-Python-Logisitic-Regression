# Our primary objective in this step is to split our data into training and test sets. The training set will be used to train the model, while the test set will be used to evaluate the model.

# Let's start by creating a pandas Series called y for the dependent variable.

```python
y = loan['Default']
```

# Then we create a pandas DataFrame called X for the independent variables.

```python
X = loan[['Income','Loan Amount']]
```

03_03 Train and Evaluate the Model

We are going to use the LogisticRegression class from the sklearn.linear_model subpackage to train our model. Let's import it.

```python
from sklearn.linear_model import LogisticRegression
```

We can now instantiate a new object called classifier from the LogisticRegression class.

```python
classifier = LogisticRegression()
```


To train a model, we pass the training data (X_train and y_train) to the fit() method of the classifier object.

```python
model = classifier.fit(X_train, y_train)
```

Recall there are 9 instances (or rows) in the test set. To predict labels for the test instances, we pass the independent variables of the test set (X_test) to predict() method of the model

```python
model.predict(X_test)
```

To output lists the predictions made by the model in the order in which the instances appear in the test set.

To evaluate how accurate our model is, we pass the test data (X_test and y_test) to the score() method od the model

```python
model.score(X_test, y_test)
```

The results tells us that our Logisitic Regression model is able to correctly predict 8 of 9 (or 89%) of the labels in the test set.add

The accuracy of a model only gives us a one-dimensional perspective of performance. To get a broader perspective, we need to generate a confusion (or error) matrix of the models performance.

To do this, we need to impoprt the confusion_matrix function from the sklearn.metrics subpakcage

```python
from sklearn.metrics import confusion_matrix
```

Then we pass the depenedent variable frp, the test test (which are the actual labels) the the models predicted labels to the confusion_matrix() function

```python
confusion_matrix(y_test, model.predict(X_test))
```

The result is a 2x2 array that shows how many instances the model predicted correctly or incorrectly as either Yes or No. This confusion matrix can be illustrated as follows:



The first row of the matrix shows that of the 4 isntances there were actually No, the model predicted 3 of them as Yes. The second row of the matrix shows that of the 5 isntances that were actually Yes, the model predicted all 5 correctly as Yes.

```python
log_odds = np.round(model.coef_[0],2)
log_odds
```

```python
pd.DataFrame({'log odds': log_odds}, index = X.columns)
```

Understandably, interpreting the coefficients in terms of log odds is a bit confusing. A more intuitive approach would be to look at them in terms of odds. Let's exponentiate the coefficients so we can interpret them in terms of odds rather than log odds:

```python
odds = np.round(np.exp(log_odds),2)
pd.DataFrame({'odds': odds}, index = X.columns)
```
